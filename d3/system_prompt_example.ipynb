```python
import os, boto3
from google.colab import userdata
```

```python
for k in ["AWS_ACCESS_KEY_ID", "AWS_SECRET_ACCESS_KEY"]:
    os.environ[k] = userdata.get(k)
os.environ["AWS_DEFAULT_REGION"] = "ap-southeast-1"
client = boto3.client(
    "bedrock-runtime")
model_id = "global.anthropic.claude-haiku-4-5-20251001-v1:0"
```

```python
def append_user_message(messages, text):
    user_message = {
        "role":"user",
        "content":[
            {
                "text":text
            }
        ]
    }
    messages.append(user_message)

def append_assistant_message(messages, text):
    assistant_message = {
        "role":"assistant",
        "content":[
            {
                "text":text
            }
        ]
    }
    messages.append(assistant_message)
```

```python
from datetime import datetime, timedelta

def get_current_datetime(date_format="%Y-%m-%d %H:%M:%S"):
    return datetime.now().strftime(date_format)
```

```python
system_prompt = "You are AWS Cloud specialist. Your job is to answer questions related to cloud hosting services using AWS Services, nothing more. All answer must contain the current datetime."

def chat(messages):
    response = client.converse(
        modelId = model_id,
        messages = messages,
        system = [{"text": system_prompt}]
    )
    return response["output"]["message"]["content"][0]["text"]
```

```python
messages = []
append_user_message(messages, "How can I host a SQL Server database on AWS?")
answer = chat(messages)
print(answer)
```
