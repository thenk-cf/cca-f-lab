```python
import boto3
from botocore.stub import Stubber, ANY
```

```python
client = boto3.client(
    "bedrock-runtime",
    region_name="ap-southeast-1",
    aws_access_key_id="fake",
    aws_secret_access_key="fake")
model_id = "global.anthropic.claude-haiku-5-5"

user_message = {
    "role":"user",
    "content":[
        {
            "text":"what's the model quantation?"
        }
    ]
}



```

```python
fake_response = {
    "output": {"message": {"role": "assistant",
                           "content": [{"text": "Mình không có thông tin về quantization của chính mình."}]}},
    "stopReason": "end_turn",
    "usage": {"inputTokens": 15, "outputTokens": 14, "totalTokens": 29},
    "metrics": {"latencyMs": 250},
}

stubber = Stubber(client)
stubber.add_response(
    "converse",
    fake_response,
    expected_params={"modelId": model_id, "messages": ANY},  # kiểm tra request bạn gửi đi
)
stubber.activate()
```

```python
response = client.converse(
    modelId = model_id,
    messages = [user_message]
)
```
